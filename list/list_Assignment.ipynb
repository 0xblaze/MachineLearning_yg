```python
def transpose_matrix(matrix):
    transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposed

matrix=[
    [3,6,1],
    [4,7,1],
    [8,6,3]
]
transposed = transpose_matrix(matrix)
print("Original MAtrix:")
for row in matrix:
    print(row)
print("Transpose MAtrix:")
for row in transposed:
    print(row)

```

```python
def rotate_list(lst,n):
    ls_1=lst[n:]+lst[:n]
    return ls_1

original_list=[1,2,3,4,5,6,7,8,9]
rotated =rotate_list(original_list,3)
print(original_list)
print(rotated)
```

```python
def list_intersection(lst,lst1):
    return [x for x in lst if x in lst1]


list1 = [2,4,6,3,26,72,16,134,5]
list2 = [5,7,3,16,7,2,6,2,55]
listfinal=list_intersection(list1,list2)
print(listfinal)
```
