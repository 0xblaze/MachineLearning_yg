```python
def map_filter(filer_func,map_func,lst):
    result=[]
    for i in lst:
        if filer_func(i):
            result.append(map_func(i))
    return result

print(map_filter(lambda x:x%2==0, lambda x:x+1,[1,2,3,4,5,6,7]))
```


### Assignment 13: Function with Context Manager

Define a function that uses a context manager to write a list of integers to a file. The function should handle any errors that occur during file operations. Test with different lists.

```python
def file_op(x,filename):
    try:
        with open(filename,'w') as f:
            for i in x:
                f.write(f"{i}\n")
    except IOError as e:
        print(f"Error occured:{e}")


file_op(['a',2,3,4,5,6],'randoom.txt')
```

```python
def testing(lst):
    num,strin,flot =[],[],[]
    for item in lst:
        if isinstance(item,int):
            num.append(item)
        elif isinstance(item,str):
            strin.append(item)
        elif isinstance(item,float):
            flot.append(item)
    return num,strin,flot


print(testing([1,2,'a',6.8,7,'teja']))

   
```
